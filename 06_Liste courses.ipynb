# Liste de courses avec menus de plusieurs lignes



```python
from IPython.display import clear_output

#### Exo docstring
# https://www.docstring.fr/formations/la-liste-de-courses/creation-des-variables-globales-1735/?module=8&session=1&course=1

LISTE = []    # majuscules constantes globales
MENU_CHOICES = ["1", "2", "3", "4", "5"]

MENU = """Choisissez parmi les 5 options suivantes :
1: Ajouter une élément à la liste,
2: Retirer un élément de la liste,
3: Afficher la liste,
4: Vider la liste,
5: Quitter. """


#### Input et son traitement des mauvaises touches 
user_choice = ""


#### Quitter
# voir avec commande sys pour quitter le programme :)
while not user_choice == "5":
    # Dans la boucle des choix

    print(MENU)
    print("Ma liste : ", LISTE)

    
    # Vérifie que la réponse est bien dans la liste MENU_CHOICES
    while not user_choice in MENU_CHOICES:
        user_choice = input(f"Votre choix : {MENU_CHOICES}")
        if not user_choice in MENU_CHOICES:
            print("Erreur d'option")
            
    #### Ajout
    if user_choice == "1":
        print("Vous avez choisi d'ajouter un élément à la liste !")
        add_liste = input("Élément à ajouter : ")
        LISTE.append(add_liste)
        
        # efface les sorties 
        clear_output(wait=True)
        
        
        # Gestion de la touche 5 (Arrêt)
        user_choice = ""
    
    #### Retrait
    if user_choice == "2":
        print("Vous avez choisi le retrait d'un élément !")
        print("Rappel de votre liste : ", LISTE)
        ote_liste = input("Entrer l'élément à retirer : ")
        if ote_liste in LISTE:
            LISTE.remove(ote_liste)
            print(ote_liste, "a été retiré.")
            print("Nouvelle liste :", LISTE)
        else:
            print(ote_liste, "n'est pas dans cette liste.")
            print("Votre liste actuelle :", LISTE)

        # efface les sorties 
        clear_output(wait=True)
        
        ## Gestion de la touche 5 (Arrêt)
        user_choice = ""
    
    #### Affichage
    if user_choice == "3":
        print("Vous avez choisi affichage de votre liste en cours !")
        if not LISTE: 
            print("Votre liste est vide")
        else:
            print("Votre liste : ", LISTE)

        # efface les sorties 
        clear_output(wait=True)
        
        ## Gestion de la touche 5 (Arrêt)
        user_choice = ""
    
    
    #### Vidage
    if user_choice == "4":
        print("Vous avez choisi effacement !")
        if not LISTE:
            print("Votre liste est déjà vide")
        else:
            print("Votre liste en cours : ", LISTE)
            user_confirmation = input("Confirmer que vous voulez supprimer les éléments o/O")
            if user_confirmation == "o" or user_confirmation == "O":
                LISTE.clear()
                print("À votre demande, la liste a été vidée")
            else:
                print("Vous n'avez pas confirmer l'opération de vidage de la liste :)")

        # efface les sorties 
        clear_output(wait=True)
        
        ## Gestion de la touche 5 (Arrêt)
        user_choice = ""

# sortie de la boucle de la touche 5
print("Vous avez choisi de quitter ! À bientôt !")

```
